```python
import tensorflow as tf
```

```python
from tensorflow.keras import layers
```

```python
conv=layers.Conv2D(filters=32,kernel_size=(3,3))
```

```python
print(conv)
```

```python
# Create a dummy image: 28 × 28 pixels, 1 colour channel
```

```python
image=tf.random.normal((1,28,28,1))
```

```python
output=conv(image)
```

```python
print("input shape:",image.shape)
print("output shape:",output.shape)
```

```python
relu=layers.ReLU()
```

```python
relu_output=relu(output)
```

```python
print("before relu=" ,output.shape)
print ("after relu=",relu_output.shape)
```

```python
pool = layers.MaxPooling2D(pool_size =(2,2))
pool_output=pool(relu_output)
```

```python
print("After ReLU    :", relu_output.shape)
print("After Pooling :", pool_output.shape)
```

```python
flatten=layers.Flatten()
flatten_output=flatten(pool_output)
```

```python
print("after pooling",pool_output.shape)
print("after flatten", flatten_output.shape)
```

```python
dense=layers.Dense(10)
dense_output=dense(flatten_output)
print("after dense:", dense_output.shape)
```
